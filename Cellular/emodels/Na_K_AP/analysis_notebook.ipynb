## Summary
In this example, we simulate a single neuron (me-model) and record its membrane potential (v) together with sodium (Na⁺) and potassium (K⁺) ionic currents from specific ion channel mechanisms. By observing NaTg.ina and K_Pst.ik during an evoked action potential, we can explore how sodium influx and potassium efflux shape the waveform of the spike.

We can now create our cell using BlueCelluLab:

```python
print(f"threshold current is {emodel_properties.threshold_current} nA.")
print(f"holding current is {emodel_properties.holding_current} nA.")
```

Here, we will record both Na⁺ and K⁺ channel currents from the same location so they can be directly compared during an AP.
We use **tools.list_mechanism_variables** function to inspect available variables and find the mechanism-scoped Na⁺/K⁺ currents.

```python
mech_vars = tools.list_mechanism_variables(cell, xs=(0.5,))
# Example: pick soma[0] at x=0.5
key = next(k for k in mech_vars if k.endswith("soma[0]"))
mech_vars[key]
```

Look for:
- Na⁺ current: often NaTg.ina, na3.ina, or similar.
- K⁺ current: often K_Pst.ik, Kv3.1.ik, etc.

Add the recordings

```python
section_obj = tools.get_section(cell, "soma[0]")

cell.add_variable_recording("NaTg.ina", section_obj, 0.5)  # Na⁺ current
cell.add_variable_recording("K_Pst.ik", section_obj, 0.5)  # K⁺ current
cell.add_variable_recording("v", section_obj, 0.5)
```

Inject a simple step current to evoke an AP and simulate

```python
stim_delay = 100  # ms
stim_dur = 1350  # ms
cell.add_step(stim_delay, stim_dur, 0.5, cell.soma, 0.5)

sim = Simulation()
sim.add_cell(cell)
sim.run(2000, dt=0.05, cvode=False)
```

Retrieve and plot the results

```python
import matplotlib.pyplot as plt

t = cell.get_time()
v_trace = cell.get_variable_recording("v", section_obj, 0.5)
ina_trace = cell.get_variable_recording("NaTg.ina", section_obj, 0.5)
ik_trace = cell.get_variable_recording("K_Pst.ik", section_obj, 0.5)

plt.figure(figsize=(8,6))

# Membrane potential
plt.subplot(3,1,1)
plt.plot(t, v_trace)
plt.ylabel("V (mV)")
plt.title("AP and Na⁺ / K⁺ currents")

# Sodium current
plt.subplot(3,1,2)
plt.plot(t, ina_trace)
plt.ylabel("Na⁺ current (nA)")

# Potassium current
plt.subplot(3,1,3)
plt.plot(t, ik_trace)
plt.ylabel("K⁺ current (nA)")
plt.xlabel("Time (ms)")

plt.tight_layout()
plt.show()
```

The top panel shows the membrane potential during a train of action potentials.

The middle panel shows the inward Na⁺ current (negative deflection) that drives the rapid AP upstroke.

The bottom panel shows the outward K⁺ current that repolarizes the membrane after each AP.
