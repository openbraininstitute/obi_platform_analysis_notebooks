At a given point in time only consider the part of the neuron that is highlighted. Do not report errors for parts that are not highlighted.

In the example above, we can see that the green surface points clearly outline an additional spine that has not been detected, as no colored mesh is displayed inside of it. Had the spine been detected properly, its root would have been attached to the highlighted part of the skeleton. Thus it is a missed spine for the highlighted part and must be reported. Please make sure to check that a missed spine really belongs to the currently highlighted part, otherwise it may be reported twice.

## Types of errors
There are six types of spine mesh reconstruction errors that can be classified. Below, we will call a "true spine" what the manual proofreader (i.e.: you) considers a spine, and "detected spine" what the algorithm has found and is displayed to you.

  - Missed spine: None of the surface of a true spine is part of any detected spine.
  - Incomplete spine: A significant portion of the surface of a true spine is not part of any spine, but the "missed spine" criterion has not been reached.
  - False positive: More than half of the surface of a detected spine is not part of a true spine.
  - Falsely extended spine: A significant portion of the surface of a detected spine is not part of a true spine, but the "false positive" criterion has not been reached.
  - Merged spine: The surface of a detected spine contains parts of more than a single true spine.
  - Split spine: The surface of a true spine contains parts of more than a single detected spine.

**In this notebook we are only concerned with the first four types of errors!**
### Rule 2:
Only count and report the errors that this notebook is concerned with. Do not attept to report the other two types. There are other notebooks for this

### Examples
Above, we have already seen an example of a missed spine. Let's look at some other modes of failure.

#### Incomplete spine
The green spine below is incomplete. Parts of its neck are missing.

![image](./incomplete_example1.png)

It can be hard to draw the line at what point a spine is incomplete. In general, we want to allow for some degree of leeway. When in doubt, ask if the detected spine mesh would still be useful for an analysis. In the example above, the part with the lowest neck diameter is missing, hence it would not be useful for calculating the neck resistance, hence it is incomplete.

#### False positive
Below, the tip of the dendrite has been incorrectly detected as a spine (yellow).

![image](./false_positive_example.png)

Note that it contains parts that actually are a spine (protruding downward). But according to our definition above, a false positive has been reached when more than 50\% of the surface are non-spine. That is certainly the case.

#### Falsely extended spine 
Below, the teal spine is incorrectly extending into the surface of the surrounding dendrite.

![image](./extended_example.png)

**NOTE**: We found that this case is often misattributed by inexperienced proofreaders. Often, something can look like a false extension, but upon more careful examination actually turns out to be part of a true spine. For example, consider the blue spine below:

![image](./extended_counter_example.png)
At first glance, it looks like it is incorrectly extending to the right. But, when one studies it carefully and from different angles, you can see that the right part is actually spine that is lying almost flat against the dendrite surface! This is really hard to show in a screen shot, so you will have to take my word for it though.

### Rule 3:
Be careful with the "extended" assessment and double and triple check it!

## Errors we are not concerned with

The green spine below should probably have been detected as two separate ones. 
**Do not attempt to report this as an error.**

![image](./merger_example.png)

Below we see an interesting example. In the center of the image, there is one mushroom spine and a forked spine. But parts of the forked spine are split from it and instead merged with the mushroom spine. It is the combination of a merge and a split error.

**However:** In this notebook you should not report this as an error. Every part of the a true spine is still part of a detected spine and the other way around. None of the four error classes is triggered.

![image](./split_and_merge.png)

Below we see that the head of one spine has been merged into a neighboring spine (in red). 

**Note**: In this notebook, the merger is irrelevant. However, we see that only a part of the spine on the left is included in the merged spine. Hence, this would be reported as a single "incomplete" spine!

Download source cell mesh

We find the cell surface mesh of the neuron and download it. This is the reference that we proofread against.

```python
# Load mesh
import pylmesh
subsample_factor = 2

mesh_path = "neuron_mesh.glb"
src_mesh_entity = morphology.source_mesh_entity(db_client=client)
client.download_file(entity_id=src_mesh_entity.id, entity_type=EMCellMesh,
                     asset_id=src_mesh_entity.assets[0], output_path=mesh_path)

mesh_obj = pylmesh.load_mesh(str(mesh_path))
vtx_surf = mesh_obj.vertices[::subsample_factor]
mesh_obj = None
vtx_surf = (numpy.array([[v.x, v.y, v.z] for v in vtx_surf]) * 1E-3).astype(numpy.float32)

sec_points = pandas.concat([
    pandas.DataFrame(sec_.points[:, :3], columns=["x", "y", "z"])
    for sec_ in m.morphology.sections
], axis=0, keys=range(len(m.morphology.sections)), names=["sec_id"]).droplevel(1)

tree = KDTree(sec_points)
_, i = tree.query(vtx_surf)

vtx_surf = pandas.DataFrame(vtx_surf, index=sec_points.index[i])
```

# Launch 3d visualization

Below, we launch the 3d visualization and error reporting widget.

Note that the code cell is quite long, you might have to do a lot of scrolling.

The instructions for what to do are above. Briefly:
  1. Select the first section assigned to you
  2. Report error counts for the highlighted part, press the button to save the result and move to the next part
  3. Continue until the end of the section is reached and select the next section
  4. Once all assigned sections have been tackled, continue with the remaining cells of the notebook.

**NOTE** There is no "undo" function. If you make a mistake, you have to re-run this cell and start from scratch! BE CAREFUL!

```python
import k3d 
import time

step = 15

n_detected = widgets.Label()
n_missing = widgets.IntSlider(description="Missing", min=0, value=0, max=10)
n_incomplete = widgets.IntSlider(description="Incomplete", min=0, value=0, max=10)
n_extended = widgets.IntSlider(description="Extended", min=0, value=0, max=10)
n_false = widgets.IntSlider(description="False pos.", min=0, value=0, max=10)

plot_face = k3d.plot(background_color=0x000000, grid_visible=False, camera_auto_fit=False)

added_spine_dict = {}
added_points_lst = []
highlight_sec_lst = []
screenshots = []
screenshot_locations = []
results = []
seg_id = [0]
sec_id = [0]
hl_spine_cols = [0xCC0000, 0x0000EE, 0x999900, 0x00DD00, 0x0099AA]

for i, sec in enumerate(m.morphology.sections):
    w = 0.1
    col = 0xDDDD00
    plot_face += k3d.line(
        vertices=sec.points[:, :3],
        color=col,
        width=w,
        opacity=0.5,
        shader="mesh"
    )
    if numpy.mod(i + 1, 100) == 0:
        time.sleep(0.1)

def add_section_info(tgt_section):
    global plot_face
    global seg_id
    global sec_id
    sec_id.clear()
    sec_id.append(tgt_section)
    tgt_sec = m.morphology.sections[tgt_section - 1]
    tgt_segment_s = 0
    tgt_segment_e = numpy.minimum(tgt_segment_s + step, len(tgt_sec.points))
    seg_id.clear()
    seg_id.append(tgt_segment_s)
    cam_tgt = tgt_sec.points[tgt_segment_s:tgt_segment_e, :3].mean(axis=0)
    cam_up = numpy.array([0, 1, 0])
    cam_cross = numpy.cross(cam_up, tgt_sec.points[-1, :3] - tgt_sec.points[0, :3])
    cam_cross = 5.0 * cam_cross / numpy.linalg.norm(cam_cross)
    cam_pos = cam_tgt + cam_cross
    cam_params = numpy.hstack([cam_pos, cam_tgt, cam_up])  

    for spn in added_spine_dict.values():
        plot_face -= spn
    added_spine_dict.clear()
    while len(added_points_lst) > 0:
        plot_face -= added_points_lst.pop()
    while len(highlight_sec_lst) > 0:
        plot_face -= highlight_sec_lst.pop()

    plot_face.camera = cam_params
    n_spine_hl = 0
    for _, row in m.spines.spine_table.set_index("afferent_section_id").loc[[tgt_section], ["spine_id", "afferent_segment_id"]].iterrows():
        spine_id = row["spine_id"]
        in_highlight = tgt_segment_s <= row["afferent_segment_id"] < tgt_segment_e
        spine_col = 0x777777
        if in_highlight:
            n_spine_hl += 1
            spine_col = hl_spine_cols.pop(0)
            hl_spine_cols.append(spine_col)
        vtx = m.spines.spine_mesh_points(spine_id).astype(numpy.float32)
        fac = numpy.asarray(m.spines.spine_mesh_triangles(spine_id), dtype=numpy.uint32)
        if len(fac) > 0:
            spine = k3d.mesh(
            vtx, fac,
            color=spine_col,
            opacity=1.0
            )
            plot_face += spine
            added_spine_dict[spine_id] = spine
    n_detected.value = f"{n_spine_hl} spines detected"
    n_incomplete.max = n_spine_hl
    n_extended.max = n_spine_hl
    n_false.max = n_spine_hl
    n_missing.value = 0
    n_incomplete.value = 0
    n_extended.value = 0
    n_false.value = 0
    
    hl_line = k3d.line(
        vertices=tgt_sec.points[tgt_segment_s:(tgt_segment_e+1), :3],
        color=0xCC3300,
        width=0.15,
        opacity=0.75,
        shader="mesh"
    )
    plot_face += hl_line
    highlight_sec_lst.append(hl_line)

    if (tgt_section - 1) in vtx_surf.index:
        added_points_lst.append(k3d.points(
        vtx_surf.loc[tgt_section - 1].to_numpy(),
        point_size=0.05,
        opacity=0.5,
        color=0x009900,
        shader="points"))
        plot_face += added_points_lst[-1]

def move_distal(ignore_):
    global plot_face
    global seg_id
    global sec_id
    tgt_section = sec_id[0]
    tgt_sec = m.morphology.sections[tgt_section - 1]
    res_dict = {"section": sec_id[0], "segment_start": seg_id[0],
                "segment_end": numpy.minimum(seg_id[0] + step, len(tgt_sec.points)),
                "detected": int(n_detected.value.split()[0]), "missing": n_missing.value,
                "incomplete": n_incomplete.value, "extended": n_extended.value, "false_positive": n_false.value}
    results.append(res_dict)
    tgt_segment_s = numpy.minimum(seg_id[0] + step, len(tgt_sec.points) - 1)
    tgt_segment_e = numpy.minimum(tgt_segment_s + step, len(tgt_sec.points))
    seg_id.append(tgt_segment_s)
    seg_id.pop(0)
    cam_tgt = tgt_sec.points[tgt_segment_s:tgt_segment_e, :3].mean(axis=0)
    cam_up = numpy.array([0, 1, 0])
    cam_cross = numpy.cross(cam_up, tgt_sec.points[-1, :3] - tgt_sec.points[0, :3])
    cam_cross = 5.0 * cam_cross / numpy.linalg.norm(cam_cross)
    cam_pos = cam_tgt + cam_cross
    cam_params = numpy.hstack([cam_pos, cam_tgt, cam_up])  
    plot_face.camera = cam_params
    spine_table_ = m.spines.spine_table.set_index("spine_id")
    n_spine_hl = 0
    for spn_id, spn in added_spine_dict.items():
        if tgt_segment_s <= spine_table_.loc[spn_id, "afferent_segment_id"] < tgt_segment_e:
            spine_col = hl_spine_cols.pop(0)
            hl_spine_cols.append(spine_col)
            spn.color = spine_col
            spn.opacity = 1.0
            n_spine_hl += 1
        else:
            spn.color = 0x777777
            spn.opacity = 0.5
    
    n_detected.value = f"{n_spine_hl} spines detected"
    n_incomplete.max = n_spine_hl
    n_extended.max = n_spine_hl
    n_false.max = n_spine_hl
    n_missing.value = 0
    n_incomplete.value = 0
    n_extended.value = 0
    n_false.value = 0

    while len(highlight_sec_lst) > 0:
        plot_face -= highlight_sec_lst.pop()
    hl_line = k3d.line(
        vertices=tgt_sec.points[tgt_segment_s:(tgt_segment_e+1), :3],
        color=0xCC3300,
        width=0.15,
        opacity=0.75,
        shader="mesh"
    )
    plot_face += hl_line
    highlight_sec_lst.append(hl_line)

def take_screenshot(ignore_):
    global plot_face
    global seg_id
    global sec_id
    screenshot_locations.append((sec_id[0], seg_id[0]))
    if len(plot_face.screenshot) > 0:
        screenshots.append(plot_face.screenshot)
    plot_face.fetch_screenshot()

sel_sec = widgets.Dropdown(options={
    f"Section {sec}: {val} spines": sec for sec, val in m.spines.spine_table["afferent_section_id"].value_counts().items()
    })
move_next = widgets.Button(description="Save and move next")
move_next.on_click(move_distal)
take_shot = widgets.Button(description="Screenshot")
take_shot.on_click(take_screenshot)
plot_face.display()
display.display(widgets.interactive(add_section_info, tgt_section=sel_sec))
display.display(widgets.VBox([
    widgets.HBox([n_detected, move_next, take_shot]),
    widgets.HBox([n_false, n_extended, n_incomplete, n_missing])
    ]))
```
