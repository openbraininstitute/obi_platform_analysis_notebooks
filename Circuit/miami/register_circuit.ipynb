## Circuit information **→ To be filled in**

### General metadata

To be provided as a dict with key/value pairs. Optional entries can be set to None.

**Linked entities must exist in entitycore. Otherwise, they must first be registered:**
- Species
- Subject
- Brain region
- License


```python
circuit_metadata = {
    # Name of the circuit
    "name": parent.name + "-AD",

    # Description of the circuit
    "description": f"Modification of the original {parent.name} with reduction of 50% of g_bar for Ih current in PC",
    
    # Scale of the circuit (single, pair, small, microcircuit, region, system, whole_brain)
    "scale": parent.scale,

    # Optional: Name of the root circuit (= top parent in the extraction/derivation hierarchy), if any
    "root": None,

    # Optional: Name of the parent circuit (=direct parent the circuit was extracted or derived from)
    "parent": None,

    # Optional: Type of circuit derivation (circuit_extraction, circuit_rewiring), if parent provided
    "derivation_type": None,

    # Species name [Must exist in entitycore; otherwise, must first be registered]
    "species": parent.subject.species.name,

    # Subject name [Must exist in entitycore; otherwise, must first be registered]
    "subject": parent.subject.name,

    # Brain region name [Must exist in entitycore; otherwise, must first be registered]
    "brain_region": parent.brain_region.name,

    # Optional: License name [Must exist in entitycore; otherwise, must first be registered]
    "license": parent.license.label,

    # Optional: Short, human readable publication string
    "published_in": parent.published_in,

    # Optional: Contact e-mail address
    "contact": parent.contact_email,

    # Optional: Experiment date (=date when the circuit has been build or published)
    # Note: Two date formats supported: "November, 2024" (treated as 01.11.2024) and "27.03.2024"
    "experiment_date": "November, 2024",

    # Circuit build category (computational_model, em_reconstruction)
    "build_category": "computational_model",

    # Properties of the circuit
    "has_morphologies": True,
    "has_point_neurons": False,
    "has_electrical_cell_models": True,
    "has_spines": False,

    "number_neurons": parent.number_neurons,  # Number of intrinsic neurons
    "number_synapses": parent.number_synapses,  # Number of intrinsic synapses; for synaptomes only: incl. extrinsic synapses
    "number_connections": parent.number_connections,  # [Optional] Number of intrinsic connections; for synaptomes only: incl. extrinsic connections
    }
```

## Check provided metadata and dependencies

```python
from register_circuit_helpers import check_if_circuit_exists, check_counts, get_contributions, get_brain_region, get_exp_date, get_license, get_parent_circuit, get_publications, get_root_circuit, get_subject
```

### Check if circuit is registered in entitycore already

The circuit must not exist in entitycore yet!

```python
check_if_circuit_exists(client, circuit_metadata)
```

### Check and retrieve dependencies

All dependencies must already exist in entitycore!

```python
# Check/get subject
subject = get_subject(client, circuit_metadata)

# Check/get brain region
# brain_region = get_brain_region(client, circuit_metadata)
# brain_region_id=UUID('f11a8cad-9792-4aa3-8f7a-e4ed92433d61') # staging
brain_region_id=UUID('b0ffdc9a-933b-4d5e-aa3d-9f0d57a92bdf')   # production
brain_region = client.get_entity(entity_id=brain_region_id, entity_type=models.BrainRegion)

# Check/get license
license = get_license(client, circuit_metadata)

# Check/get root circuit
root = get_root_circuit(client, circuit_metadata)

# Check/get parent circuit [Note: Parent = Derivation of type <derivation_type>]
parent = get_parent_circuit(client, circuit_metadata)

# Check neuron/synapse/connection counts
check_counts(circuit_metadata)

# Determine experimental data
# Note: Two date formats supported: "November, 2024" (treated as 01.11.2024) and "27.03.2024"
exp_date = get_exp_date(circuit_metadata)

# Check/get contributing agents
contribution_dict = get_contributions(client, circuit_contributions, verbose=False)

# Check/get publications
publication_dict = get_publications(client, circuit_publications, verbose=False)
```

## Register circuit entity, assets, and links

- Use `check_only` for a dry run without actual registration first!
- Use `make_public` to register as a public entity

```python
check_only = False  # If True, only check, no actual registration
make_public = False  # Register as public entity
```

```python
from register_circuit_helpers import register_circuit_entity, register_asset, register_contributions, register_derivation, register_publication_links
```

### Register circuit entity

```python
# Create & register circuit entity
registered_circuit = register_circuit_entity(client, circuit_metadata, subject, brain_region, license, root, exp_date, make_public=make_public, check_only=check_only)
```

### Register assets
- SONATA circuit folder
- Optional: Overview/sim designer image
- Optional: Compressed SONATA circuit
- Optional: Connectivity matrix folder
- Optional: Basic connectivity figures


```python
# Register circuit folder asset
circuit_folder_asset = register_asset(client, file_path=circuit_path, asset_label="sonata_circuit", registered_circuit=registered_circuit, check_only=check_only)

# Register compressed circuit asset
gz_asset = register_asset(client, file_path=compressed_circuit_path, asset_label="compressed_sonata_circuit", registered_circuit=registered_circuit, check_only=check_only)

# Connectivity matrix folder asset
matrix_folder_asset = register_asset(client, file_path=matrix_path, asset_label="circuit_connectivity_matrices", registered_circuit=registered_circuit, check_only=check_only)

# Main circuit figure for detailed explore page
main_fig_asset = register_asset(client, file_path=main_figure_path, asset_label="circuit_visualization", registered_circuit=registered_circuit, check_only=check_only)

# Circuit figure for simulation designer
sim_fig_asset = register_asset(client, file_path=sim_designer_figure_path, asset_label="simulation_designer_image", registered_circuit=registered_circuit, check_only=check_only)

# Basic connectivity plots
node_stats_fig_asset = register_asset(client, file_path=node_stats_figure_path, asset_label="node_stats", registered_circuit=registered_circuit, check_only=check_only)
network_stats_a_fig_asset = register_asset(client, file_path=network_stats_a_figure_path, asset_label="network_stats_a", registered_circuit=registered_circuit, check_only=check_only)
network_stats_b_fig_asset = register_asset(client, file_path=network_stats_b_figure_path, asset_label="network_stats_b", registered_circuit=registered_circuit, check_only=check_only)
```

### Register entity links
- Derivation links (if any)
- Contribution links (if any)
- Publication links (if any)

```python
# Derivation links
derivation = register_derivation(client, from_entity=parent, derivation_type=circuit_metadata.get("derivation_type"), registered_circuit=registered_circuit, check_only=check_only)

# Contribution links
contributions = register_contributions(client, contribution_dict=contribution_dict, registered_circuit=registered_circuit, check_only=check_only)
    
# Publication links
publ_links = register_publication_links(client, publication_dict=publication_dict, registered_circuit=registered_circuit, check_only=check_only)
```
