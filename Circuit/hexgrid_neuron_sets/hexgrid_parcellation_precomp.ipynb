# Hexgrid parcellation (pre-computed)

Using pre-computed parcellation from [ConnectomeUtilities](https://github.com/openbraininstitute/ConnectomeUtilities/blob/main/examples/data/voxel-based-hex-grid-info-with-conicality.h5) repo.

Requirements:
- conntility
- plotly
- kaleido

```python
import bluepysnap as snap
import json
import numpy as np
import pandas
import plotly.express as px
import plotly.graph_objects as go
import random

from conntility.circuit_models import neuron_groups
from pathlib import Path
```

```python
circ_fn = "/Users/pokorny/Data/Circuits/nbS1-partial/circuit_config.json"
column_prop_fn = "./voxel-based-hex-grid-info-with-conicality.h5"
```

```python
# Load circuit & base grouping
circ = snap.Circuit(circ_fn)
popul = "S1nonbarrel_neurons"
loader_config = {
    "loading": {
        "base_target": "All", 
        "node_population": popul,
        "properties": ["x", "y", "z", "etype", "mtype", "layer", "synapse_class"],
        "atlas": [
            {"data": "./column_identities.nrrd",
             "properties": ["column_id"]}
        ]
    },
    # "filtering":[
    #     {
    #         "column": "layer",
    #         "value": "4"
    #     }
    # ],
    "grouping": [
        {
            "method": "group_by_properties",
            "columns": ["column_id"]
        }
    ]
}

base_grp = neuron_groups.load_group_filter(circ, loader_config)

display(base_grp)
# NOTE: column_id 0 indicates a node is NOT MEMBER OF ANY COLUMN!
```

```python
# Load hexgrid column properties
col_props = pandas.read_hdf(column_prop_fn, "grid-info").set_index("nrrd-file-id")
display(col_props)
```

```python
# Select hex subtargets

## No selection --> Show all
# subtarget_sel = None

## Manual selection
# subtarget_sel = ["R18;C1", "R17;C1"]

## Long slice selection
R = [str(r) for r in range(18, 0, -1)]  # Rows 18, 17, ..., 1
C = np.repeat([str(c) for c in range(1, 10, 1)], 2)  # Columns 1, 1, 2, 2, ..., 9, 9
subtarget_sel = [f"R{R};C{C}" for R, C in zip(R, C)]

if subtarget_sel:
    col_props_sel = col_props.set_index("grid-subtarget")
    col_props_sel["column_id"] = col_props.index
    col_props_sel = col_props_sel.loc[subtarget_sel]
    print(f"{col_props_sel.shape[0]} columns selected: {col_props_sel["column_id"].values}")
else:
    col_props_sel = col_props
    print(f"All {len(col_props_sel.index)} columns selected: {col_props_sel.index.to_numpy()[0]}..{col_props_sel.index.to_numpy()[-1]}")
```

```python
# Plot 2D grid
fig = go.Figure()

if subtarget_sel:
    fig.add_trace(go.Scatter(
        x=col_props["grid-x"],
        y=col_props["grid-y"],
        mode="markers",
        marker=dict(symbol="hexagon", size=35, color="cornflowerblue", opacity=0.2),
    ))

fig.add_trace(go.Scatter(
    x=col_props_sel["grid-x"],
    y=col_props_sel["grid-y"],
    mode="markers+text",
    marker=dict(symbol="hexagon", size=35, color="cornflowerblue", opacity=1.0),
    text=col_props_sel.index,
    textposition="middle center",
    textfont=dict(size=7, color="black"),
))

fig.update_layout(
    width=800,
    height=800,
    xaxis=dict(visible=False),
    yaxis=dict(visible=False, scaleanchor="x"),
    title=dict(text="2D Hex Grid with Subtargets", x=0.5, xanchor="center", y=0.87),
    showlegend=False,
    plot_bgcolor="white",
    paper_bgcolor="white",
)

fig.show()
fig.write_image("2d_hexgrid.png", scale=2)
```

```python
# 3D plot of hex grid pracellation
base_grp_plot = base_grp.sample(frac=0.1, random_state=0)  # Subsample for plotting
base_grp_plot["column_id_category"] = base_grp_plot["column_id"].astype(str)  # To be treated as categorical column

colors = list(px.colors.qualitative.Alphabet)
random.seed(0)  # Change seed for different randomization
random.shuffle(colors)

fig = go.Figure()

if subtarget_sel:
    fig = px.scatter_3d(base_grp_plot, x="x", y="y", z="z", color="column_id_category", opacity=0.02)
    fig.update_traces(marker=dict(color="grey"))
    base_grp_plot_sel = base_grp_plot[np.isin(base_grp_plot["column_id"], col_props_sel["column_id"])]
    fig_sel = px.scatter_3d(base_grp_plot_sel, x="x", y="y", z="z", color="column_id_category", opacity=1.0, color_discrete_sequence=colors)
    for trace in fig_sel.data:
        fig.add_trace(trace)
else:
    fig = px.scatter_3d(base_grp_plot, x="x", y="y", z="z", color="column_id_category", opacity=1.0, color_discrete_sequence=colors)
fig.update_traces(marker=dict(size=1))
fig.update_layout(
    width=800,
    height=800,
    title=dict(text="3D Hex Grid Parcellation", x=0.5, xanchor="center", y=0.87),
    showlegend=False
    )

fig.show()
fig.write_image("3d_hexgrid.png", scale=2)
```

### Getting the actual neuron IDs

```python
# Get all selected neurons
base_grp_sel = base_grp.loc[np.isin(base_grp["column_id"], col_props_sel["column_id"])]
print(f"Total selected count: {base_grp_sel.shape[0]} neurons")
if subtarget_sel:
    print(f"Average count per column: {base_grp_sel.shape[0]/len(subtarget_sel):.1f} neurons")

# Write to .json file
node_set_dict = {
    "custom_node_set": {
        "population": popul,
        "node_id": base_grp_sel["node_ids"].to_list(),
        }
    }
with Path("node_sets.json").open("w") as f:
    json.dump(node_set_dict, f, indent=None)
```

```python
# Neuron counts per column
neuron_counts = base_grp_sel[["node_ids", "column_id"]].groupby("column_id").count()
neuron_counts.columns = ["neuron-count"]
neuron_counts["grid-subtarget"] = col_props["grid-subtarget"].loc[neuron_counts.index]
neuron_counts = neuron_counts.loc[col_props_sel["column_id"]]
neuron_counts
```

```python
# Plot neuron counts per column
fig = px.bar(neuron_counts.reset_index(), x="grid-subtarget", y="neuron-count")
fig.update_layout(
    title=dict(text="Neuron Count per Subtarget", x=0.5, xanchor="center"),
    xaxis_title="Subtarget",
    yaxis_title="Neuron count",
)
fig.show()
fig.write_image("neuron_counts.png", scale=4)
```
